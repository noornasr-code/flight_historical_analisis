```sql
SELECT

    COUNT(*) AS total_records,

    SUM(CASE WHEN ArrDelay IS NULL THEN 1 ELSE 0 END) AS ArrDelay_nulls,

    SUM(CASE WHEN DepDelay IS NULL THEN 1 ELSE 0 END) AS DepDelay_nulls,

    SUM(CASE WHEN CarrierDelay IS NULL THEN 1 ELSE 0 END) AS CarrierDelay_nulls,

    SUM(CASE WHEN WeatherDelay IS NULL THEN 1 ELSE 0 END) AS WeatherDelay_nulls,

    SUM(CASE WHEN NASDelay IS NULL THEN 1 ELSE 0 END) AS NASDelay_nulls,

    SUM(CASE WHEN SecurityDelay IS NULL THEN 1 ELSE 0 END) AS SecurityDelay_nulls,

    SUM(CASE WHEN LateAircraftDelay IS NULL THEN 1 ELSE 0 END) AS LateAircraftDelay_nulls

FROM workspace.default.flight_data_200k;
```

```sql
SELECT

    COUNT(*) AS total_records,

    COUNT(DISTINCT

        CONCAT(

            Year, '-', Month, '-', DayofMonth, '-',

            FlightNum, '-', Origin, '-', Dest, '-', TailNum

        )

    ) AS unique_records

FROM workspace.default.flight_data_200k;
```

```sql
DROP TABLE IF EXISTS workspace.default.flight_data_cleaned;
```

```sql
CREATE TABLE workspace.default.flight_data_cleaned AS

SELECT

    ActualElapsedTime,

    AirTime,

    ArrDelay,

    ArrTime,

    CRSArrTime,

    CRSDepTime,

    CRSElapsedTime,

    CancellationCode,

    Cancelled,

    CarrierDelay,

    DayOfWeek,

    DayofMonth,

    DepDelay,

    DepTime,

    Dest,

    Distance,

    Diverted,

    FlightNum,

    LateAircraftDelay,

    Month,

    NASDelay,

    Origin,

    SecurityDelay,

    TailNum,

    TaxiIn,

    TaxiOut,

    UniqueCarrier,

    WeatherDelay,

    Year

FROM workspace.default.flight_data_200k

WHERE

    Origin IS NOT NULL

    AND Dest IS NOT NULL

    AND UniqueCarrier IS NOT NULL

    AND FlightNum IS NOT NULL;
```

```sql
SELECT COUNT(*) AS total_cleaned_records

FROM workspace.default.flight_data_cleaned;
```

```sql
SELECT

    UniqueCarrier AS airline,

    COUNT(*) AS total_flights,

    ROUND(AVG(ArrDelay), 2) AS avg_arrival_delay,

    ROUND(AVG(DepDelay), 2) AS avg_departure_delay

FROM workspace.default.flight_data_cleaned

GROUP BY UniqueCarrier

ORDER BY avg_arrival_delay DESC;
```

```sql
SELECT

    UniqueCarrier AS airline,

    COUNT(*) AS total_flights,

    ROUND(AVG(try_cast(ArrDelay AS DOUBLE)), 2) AS avg_arrival_delay,

    ROUND(AVG(try_cast(DepDelay AS DOUBLE)), 2) AS avg_departure_delay

FROM workspace.default.flight_data_cleaned

GROUP BY UniqueCarrier

ORDER BY avg_arrival_delay DESC;
```

```sql
SELECT

    ROUND(SUM(CarrierDelay), 2) AS carrier_delay,

    ROUND(SUM(WeatherDelay), 2) AS weather_delay,

    ROUND(SUM(NASDelay), 2) AS nas_delay,

    ROUND(SUM(SecurityDelay), 2) AS security_delay,

    ROUND(SUM(LateAircraftDelay), 2) AS late_aircraft_delay

FROM workspace.default.flight_data_cleaned;
```

```sql
SELECT

    ROUND(SUM(try_cast(CarrierDelay AS DOUBLE)), 2) AS carrier_delay,

    ROUND(SUM(try_cast(WeatherDelay AS DOUBLE)), 2) AS weather_delay,

    ROUND(SUM(try_cast(NASDelay AS DOUBLE)), 2) AS nas_delay,

    ROUND(SUM(try_cast(SecurityDelay AS DOUBLE)), 2) AS security_delay,

    ROUND(SUM(try_cast(LateAircraftDelay AS DOUBLE)), 2) AS late_aircraft_delay

FROM workspace.default.flight_data_cleaned;
```

```sql
SELECT
    'Carrier Delay' AS delay_reason,
    SUM(try_cast(CarrierDelay AS DOUBLE)) AS total_delay
FROM workspace.default.flight_data_cleaned

UNION ALL

SELECT
    'Weather Delay',
    SUM(try_cast(WeatherDelay AS DOUBLE))
FROM workspace.default.flight_data_cleaned

UNION ALL

SELECT
    'NAS Delay',
    SUM(try_cast(NASDelay AS DOUBLE))
FROM workspace.default.flight_data_cleaned

UNION ALL

SELECT
    'Security Delay',
    SUM(try_cast(SecurityDelay AS DOUBLE))
FROM workspace.default.flight_data_cleaned

UNION ALL

SELECT
    'Late Aircraft Delay',
    SUM(try_cast(LateAircraftDelay AS DOUBLE))
FROM workspace.default.flight_data_cleaned

ORDER BY total_delay DESC;
```

```sql
SELECT

    COUNT(*) AS total_flights,

    SUM(

        CASE

            WHEN try_cast(ArrDelay AS DOUBLE) > 0 THEN 1

            ELSE 0

        END

    ) AS delayed_flights,

    ROUND(

        SUM(

            CASE

                WHEN try_cast(ArrDelay AS DOUBLE) > 0 THEN 1

                ELSE 0

            END

        ) * 100.0 / COUNT(*),

        2

    ) AS delay_percentage

FROM workspace.default.flight_data_cleaned;
```

```sql
SELECT

    Month,

    COUNT(*) AS total_flights,

    ROUND(

        AVG(try_cast(ArrDelay AS DOUBLE)),

        2

    ) AS avg_arrival_delay,

    ROUND(

        AVG(try_cast(DepDelay AS DOUBLE)),

        2

    ) AS avg_departure_delay,

    SUM(

        CASE

            WHEN try_cast(ArrDelay AS DOUBLE) > 0 THEN 1

            ELSE 0

        END

    ) AS delayed_flights

FROM workspace.default.flight_data_cleaned

GROUP BY Month

ORDER BY Month;
```

```sql
SELECT

    DayOfWeek,

    COUNT(*) AS total_flights,

    ROUND(

        AVG(try_cast(ArrDelay AS DOUBLE)),

        2

    ) AS avg_arrival_delay,

    ROUND(

        AVG(try_cast(DepDelay AS DOUBLE)),

        2

    ) AS avg_departure_delay

FROM workspace.default.flight_data_cleaned

GROUP BY DayOfWeek

ORDER BY DayOfWeek;
```

```sql
SELECT

    Year,

    COUNT(*) AS total_flights,

    ROUND(

        AVG(try_cast(ArrDelay AS DOUBLE)),

        2

    ) AS avg_arrival_delay,

    ROUND(

        AVG(try_cast(DepDelay AS DOUBLE)),

        2

    ) AS avg_departure_delay,

    SUM(

        CASE

            WHEN try_cast(ArrDelay AS DOUBLE) > 0 THEN 1

            ELSE 0

        END

    ) AS delayed_flights

FROM workspace.default.flight_data_cleaned

GROUP BY Year

ORDER BY Year;
```

```sql
SELECT

    UniqueCarrier AS airline,

    COUNT(*) AS total_flights,

    ROUND(AVG(try_cast(ArrDelay AS DOUBLE)), 2) AS avg_arrival_delay

FROM workspace.default.flight_data_cleaned

GROUP BY UniqueCarrier

HAVING COUNT(*) >= 100

ORDER BY avg_arrival_delay DESC

LIMIT 10;
```

```sql
SELECT

    Month,

    ROUND(AVG(try_cast(ArrDelay AS DOUBLE)), 2) AS avg_arrival_delay,

    SUM(

        CASE

            WHEN try_cast(ArrDelay AS DOUBLE) > 0 THEN 1

            ELSE 0

        END

    ) AS delayed_flights

FROM workspace.default.flight_data_cleaned

GROUP BY Month

ORDER BY avg_arrival_delay DESC;
```
